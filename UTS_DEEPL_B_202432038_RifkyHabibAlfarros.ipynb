Data sunspot bulanan merupakan deret waktu (time-series) yang bersifat non-linear, berfluktuasi, dan memiliki pola periodik yang amplitudonya tidak seragam dari satu siklus ke siklus lain. Metode statistik klasik kesulitan menangkap karakteristik ini, sedangkan pendekatan Deep Learning berbasis sequence mampu mempelajari ketergantungan antar-waktu secara otomatis dari data historis. LSTM mengatasi hal ini melalui cell state serta mekanisme forget gate, input gate, dan output gate yang mengatur informasi mana yang disimpan atau dibuang. Kemampuan ini cocok untuk data sunspot yang polanya berulang dalam siklus panjang dan memerlukan memori jangka panjang dalam jendela pengamatan 24 bulan.

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432038_RifkyHabibAlfarros
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Reproducibility: seed tetap agar hasil dapat diulang
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

# Gunakan GPU jika tersedia
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device yang digunakan:', device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432038_RifkyHabibAlfarros
# 1. Muat dataset (file Sunspots.csv harus berada di folder yang sama dengan notebook)
df = pd.read_csv('Sunspots.csv')

# Buang kolom indeks bawaan, ubah nama kolom, dan ubah Date menjadi tipe datetime
df = df.drop(columns=['Unnamed: 0'])
df.columns = ['Date', 'Sunspots']
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').reset_index(drop=True)

print('Jumlah baris & kolom :', df.shape)
print('Rentang waktu        :', df['Date'].min().date(), 's/d', df['Date'].max().date())
print('Missing value        :\n', df.isnull().sum())
display(df.head())
display(df.describe())

# Eksplorasi: plot seluruh deret waktu
plt.figure(figsize=(12, 4))
plt.plot(df['Date'], df['Sunspots'], color='#114F6A', linewidth=1)
plt.title('Monthly Mean Total Sunspot Number (1749 - 2021)')
plt.xlabel('Tahun')
plt.ylabel('Jumlah Sunspot')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 2. Pembagian data latih & uji secara kronologis (80% : 20%), TANPA diacak
data = df['Sunspots'].values.reshape(-1, 1).astype(np.float32)
split_idx = int(len(data) * 0.8)
train_data, test_data = data[:split_idx], data[split_idx:]
print('Data latih:', train_data.shape, '| Data uji:', test_data.shape)

# Normalisasi Min-Max (0-1). Min/Max dihitung dari data latih saja agar tidak data leakage
data_min, data_max = train_data.min(), train_data.max()
train_scaled = (train_data - data_min) / (data_max - data_min)
test_scaled  = (test_data  - data_min) / (data_max - data_min)

# 3. Sliding window: 24 bulan sebelumnya -> memprediksi bulan berikutnya
WINDOW = 24

def create_sequences(series, window):
    X, y = [], []
    for i in range(len(series) - window):
        X.append(series[i:i + window])
        y.append(series[i + window])
    return np.array(X), np.array(y)

X_train, y_train = create_sequences(train_scaled, WINDOW)
# Data uji diberi 'WINDOW' titik terakhir dari data latih sebagai konteks awal
test_with_context = np.concatenate([train_scaled[-WINDOW:], test_scaled])
X_test, y_test = create_sequences(test_with_context, WINDOW)

# Konversi ke PyTorch Tensor, bentuk X: (jumlah_sampel, window, 1)
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)
X_test  = torch.tensor(X_test,  dtype=torch.float32)
y_test  = torch.tensor(y_test,  dtype=torch.float32)

print('X_train:', X_train.shape, '| y_train:', y_train.shape)
print('X_test :', X_test.shape,  '| y_test :', y_test.shape)

# DataLoader untuk training per mini-batch (data latih boleh diacak antar-window, aman karena tiap window sudah utuh)
BATCH_SIZE = 32
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432038_RifkyHabibAlfarros
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, num_layers=2, output_size=1, dropout=0.2):
        super(SequenceModel, self).__init__()
        # Layer LSTM: input (batch, window, 1) -> output (batch, window, hidden_size)
        self.lstm = nn.LSTM(input_size=input_size,
                            hidden_size=hidden_size,
                            num_layers=num_layers,
                            batch_first=True,
                            dropout=dropout)
        # Layer Linear: memetakan hidden state terakhir menjadi 1 nilai prediksi
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        out, (h_n, c_n) = self.lstm(x)   # out: seluruh hidden state tiap time step
        last_out = out[:, -1, :]          # ambil hidden state pada time step terakhir
        return self.fc(last_out)          # prediksi bulan berikutnya
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432038_RifkyHabibAlfarros
# 1. Inisiasi hyperparameter
epochs = 50
learning_rate = 0.001
hidden_size = 64
num_layers = 2

# 2. Inisiasi model, fungsi Loss, dan Optimizer
model = SequenceModel(input_size=1, hidden_size=hidden_size, num_layers=num_layers).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
print(model)

# 3. Simpan riwayat nilai loss untuk divisualisasikan
train_losses = []

# 4. Training loop
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)

        optimizer.zero_grad()          # reset gradien
        pred = model(xb)               # forward pass
        loss = criterion(pred, yb)     # hitung MSE
        loss.backward()                # backpropagation
        optimizer.step()               # update bobot

        running_loss += loss.item() * xb.size(0)

    epoch_loss = running_loss / len(train_loader.dataset)
    train_losses.append(epoch_loss)
    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f'Epoch [{epoch+1:>2}/{epochs}]  Loss: {epoch_loss:.6f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432038_RifkyHabibAlfarros
# 1. Plot Grafik Penurunan Loss Training terhadap Epoch
plt.figure(figsize=(10, 4))
plt.plot(range(1, epochs + 1), train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Prediksi pada data uji
model.eval()
with torch.no_grad():
    y_pred_scaled = model(X_test.to(device)).cpu().numpy()

# Kembalikan ke skala asli (inverse Min-Max)
y_pred = y_pred_scaled * (data_max - data_min) + data_min
y_true = y_test.numpy() * (data_max - data_min) + data_min

# Metrik evaluasi
mse  = np.mean((y_true - y_pred) ** 2)
rmse = np.sqrt(mse)
mae  = np.mean(np.abs(y_true - y_pred))
r2   = 1 - np.sum((y_true - y_pred) ** 2) / np.sum((y_true - y_true.mean()) ** 2)
print(f'MSE  : {mse:.4f}')
print(f'RMSE : {rmse:.4f}')
print(f'MAE  : {mae:.4f}')
print(f'R2   : {r2:.4f}')

# 3. Plot Grafik Perbandingan Data Aktual vs Hasil Prediksi Model
test_dates = df['Date'].values[split_idx:]
plt.figure(figsize=(12, 5))
plt.plot(test_dates, y_true, color='#114F6A', linewidth=1.5, label='Data Aktual')
plt.plot(test_dates, y_pred, color='#1FA4B1', linewidth=1.5, label='Prediksi LSTM')
plt.title('Perbandingan Data Aktual vs Prediksi pada Data Uji')
plt.xlabel('Tahun')
plt.ylabel('Jumlah Sunspot Bulanan')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
